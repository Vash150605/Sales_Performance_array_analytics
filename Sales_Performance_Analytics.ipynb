```python
df = pd.read_csv("sales_performance.csv")
```

```python
df = pd.read_csv("sales_performance.csv")
print(df.shape)
print(df.columns)
print(df.head())
```

```python
monthly_revenue = sum(df['revenue'])
monthly_revenue
```

```python
# Monthly total revenue array (1D)
monthly_revenue = df.groupby('month')['revenue'].sum()
monthly_revenue = np.array(monthly_revenue)
print(monthly_revenue[:12])
```

```python
#Calculate Mean, Median, Std
mean_rev = np.mean(monthly_revenue)
median_rev = np.median(monthly_revenue)
std_rev = np.std(monthly_revenue)
print("Mean Revenue:", mean_rev)
print("Median Revenue:", median_rev)
print("Std Revenue:", std_rev)
```

```python
#Find Best and Worst Month
best_month = np.argmax(monthly_revenue)
worst_month = np.argmin(monthly_revenue)
print("Highest revenue:", monthly_revenue[best_month])
print("Lowest revenue:", monthly_revenue[worst_month])
```

```python
#Quarterly Analysis
quarters = monthly_revenue.reshape(8, 3)
quarter_revenue = np.sum(quarters, axis=1)
print("Revenue of all 8 quarters:", quarter_revenue)
best_quarter = np.argmax(quarter_revenue)
print("Highest quarter revenue:", quarter_revenue[best_quarter])
```

```python
#Create Product Revenue Matrix
product_matrix = df.pivot_table(index='product_id', columns='month', values='revenue').values
print("Product Matrix Shape:", product_matrix.shape)
```

```python
#Apply Discount using Broadcasting
discount = 0.10
discounted_matrix = product_matrix * (1 - discount)
print("Original total revenue:", np.sum(product_matrix))
print("After 10% discount:", np.sum(discounted_matrix))
```

```python
#Final Summary
print("=== Project Summary ===")
print("1. Mean Monthly Revenue:", np.mean(monthly_revenue))
print("2. Best Quarter Revenue:", np.max(quarter_revenue))
print("3. Total Revenue:", np.sum(monthly_revenue))
print("4. Revenue after 10% discount:", np.sum(discounted_matrix))
```
